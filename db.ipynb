```python
from google import genai
from pymongo import MongoClient
import pandas as pd

# -------------------------
# Gemini
# -------------------------
GEMINI_API_KEY = "AQ.Ab8RN6LbG9CZQcANjtpFu_6C6VwO7ZBOXhv7Z_y3NFZ0wQoKdg"

gemini = genai.Client(api_key=GEMINI_API_KEY)


# -------------------------
# MongoDB
# -------------------------
MONGO_URI = "mongodb://localhost:27017/"

mongo_client = MongoClient(MONGO_URI)

db = mongo_client["competition_crusher"]
collection = db["internal_products"]

# -------------------------
# Ask Gemini
# -------------------------
response = gemini.models.generate_content(
    model="gemini-2.5-flash",
    contents=pd.DataFrame(collection.find()).to_dict(orient="records"),
)

answer = response.text

print(answer)


# -------------------------
# Save Gemini response
# -------------------------
document = {
    "question": "What are the features of this product?",
    "answer": answer
}

result = collection.insert_one(document)

print("Saved to MongoDB:", result.inserted_id)
```
